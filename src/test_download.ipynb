```python
import os
import urllib.request
import shutil
import tempfile
from pathlib import Path
import yaml

# Giả sử BASE_URL và FILES đã được định nghĩa ở đâu đó trong file của bạn
BASE_URL = "https://data.insideairbnb.com"
FILES = {
    "listings": "data/listings.csv.gz",
    "calendar": "data/calendar.csv.gz",
    "reviews": "data/reviews.csv.gz"
}

def load_config(path: str) -> dict:
    with open(path, encoding="utf-8") as f:
        day = yaml.safe_load(f)
    day["snapshots"] = [str(s) for s in day["snapshots"]]
    return day

def download_all(day: dict, raw_dir: str):
    data_root = day["data_root"]
    city = day["city"]
    snapshots = day["snapshots"]
    
    # b sẽ chứa danh sách các list link cho từng snapshot
    b = []
    for snapshot in snapshots:
        a = []
        for file_name, rel in FILES.items():
            # Tạo link
            link = f"{BASE_URL}/{data_root}/{snapshot}/{rel}"
            # Tạo đường dẫn file đích
            # Ví dụ: data/raw/new-york-city/2025-09-01/listings.csv.gz
            dest = Path(raw_dir) / city / snapshot / Path(rel).name
            a.append((link, dest))
        b.append(a)
        
    # 2. Duyệt qua từng snapshot (từng list trong b) để tải
    for i, snapshot_links in enumerate(b):
        snapshot = snapshots[i]
        print(f"--- Đang xử lý mốc: {snapshot} ---")
        
        for link, dest in snapshot_links:
            # Tạo thư mục cho mốc này (nếu chưa có)
            dest.parent.mkdir(parents=True, exist_ok=True)
            
            print(f"Đang tải: {dest.name} ...")
            
            # Tải file bằng urllib (theo tài liệu bạn đã đọc)
            req = urllib.request.Request(link, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(req) as response:
                with tempfile.NamedTemporaryFile(delete=False, dir=dest.parent) as tmp_file:
                    shutil.copyfileobj(response, tmp_file)
            
            # Đổi tên file tạm thành file đích
            Path(tmp_file.name).replace(dest)
            print(f"  -> Đã lưu: {dest}")

# Cách gọi hàm:
# day = load_config("configs/newyork.yaml")
# download_all(day, "data/raw")
```
