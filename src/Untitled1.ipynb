```python
def load_ids(path: str, ma: str, id_col: str, flag_col: str | None = None):
    p = Path(path)
    df = pd.read_csv(p, dtype=str, encoding="utf-8")
    df = df[df["Mã dữ liệu"].str.strip() == ma]
    if flag_col:
        f = df[flag_col].fillna("").astype(str).str.strip().str.lower()
        df = df[~f.isin(["", "0", "false", "no", "nan"])]
    s = df[id_col].dropna().astype(str).str.strip()
    # Kiểm tra nếu rỗng thì báo lỗi
    if s.empty:
        raise ValueError(f"Không có dòng nào cho '{ma}' (cột {flag_col or id_col}) trong {p}")
    return set(s)
    
```
